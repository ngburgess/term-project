```python
import cv2
import numpy as np
```

```python
img = cv2.imread("peppers.tif", cv2.IMREAD_GRAYSCALE)
cv2.imwrite("peppers_clean.pgm", img)
```

```python
img = cv2.imread("peppers_clean.pgm", cv2.IMREAD_GRAYSCALE)
img_float = img.astype(np.float32)

mean = 0
sigma_values = [10, 20, 40]

for sigma in sigma_values:
    noise = np.random.normal(mean, sigma, img.shape)

    noisy = img_float + noise
    noisy = np.clip(noisy, 0, 255)
    noisy = noisy.astype(np.uint8)

    cv2.imwrite(f"peppers_gaussian_{sigma}.pgm", noisy)
```

```python
img = cv2.imread("peppers_clean.pgm", cv2.IMREAD_GRAYSCALE)

probabilities = [0.01, 0.05, 0.10]

for probability in probabilities:
    noisy = img.copy()

    rand = np.random.rand(*img.shape)

    noisy[rand < probability / 2] = 0 # pepper
    noisy[rand > 1 - probability / 2] = 255 # salt
    
    cv2.imwrite(f"peppers_salt_pepper_{probability}.pgm", noisy)
```
